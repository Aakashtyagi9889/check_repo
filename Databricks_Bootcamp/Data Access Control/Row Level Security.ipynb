### **Mapping Table**

```python
%sql
create table databricksansh.silver.rls_mapping 
(
    user_id string,
    region string
)
```

```python
%sql
insert into databricksansh.silver.rls_mapping values
("aakashtyagi9889@gmail.com" , "east"),
("avi@gmail.com" , "west"),
("ansh@gmail.com" , "south"),
("arjun@gmail.com" , "north")
```

```python
%sql
select exists
(
select * from databricksansh.silver.rls_mapping
where user_id = current_user() and 
region = "east"
)
```

### **CREATE RLS FUNCTION**

```python
%sql
CREATE OR REPLACE FUNCTION databricksansh.silver.rls_func(p_region STRING)
RETURNS BOOLEAN
RETURN
EXISTS
(
select * from databricksansh.silver.rls_mapping  
where user_id = current_user() and 
region = lower(p_region)
)
```

### **APPLYING FUNCTION ON REGION COLUMN**

```python
%sql
ALTER TABLE databricksansh.silver.stores_enr
SET ROW FILTER databricksansh.silver.rls_func ON (region)
```
