```python
data = [(1,100),(2,200),(3,300)]
schema = "id INT , salary INT"
```

```python
df =  spark.createDataFrame(data,schema)
display(df)
```

```python
df.write.format("delta")\
    .mode("append")\
    .save("/Volumes/databricksansh/bronze/bronze_volume/deltaOptimization/")
```
