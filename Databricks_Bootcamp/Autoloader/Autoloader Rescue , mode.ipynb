# **Autoloader Query**

```python
df = spark.readStream.format("cloudFiles")\
    .option("cloudFiles.format", "csv")\
    .option("cloudFiles.schemaLocation","/Volumes/databricksansh/bronze/autovol/destination/checkpoint/")\
    .option("cloudFiles.schemaEvolutionMode","rescue")\
    .load("/Volumes/databricksansh/bronze/autovol/raw/")
```

```python
df.writeStream.format("delta")\
    .outputMode("append")\
    .option("checkpointlocation","/Volumes/databricksansh/bronze/autovol/destination/checkpoint/")\
    .trigger(once=True)\
    .start("/Volumes/databricksansh/bronze/autovol/destination/data/")
```

```python
df = spark.read.format("delta")\
    .load("/Volumes/databricksansh/bronze/autovol/destination/data/")
display(df)
```

```python
df.withColumn("Rescued",col("_rescued_data")["discount"]).display()
```

```python
rescued_schema = StructType()\
    .add("discount",StringType())\
    .add("payment_method",StringType())

# Parse the stringy JSON Column 
df = df.withColumn("rescued_struct", from_json(col("_rescued_data"),rescued_schema))





```

```python
df.display()
```

```python
# Extract Individual Value 
df = df.withColumn("rescued_discount",col("rescued_struct")["discount"])\
    .withColumn("rescued_payment_method",col("rescued_struct")["payment_method"])
```
