### **Scalar Function**

### SQL

```python
%sql
CREATE OR REPLACE FUNCTION databricksansh.bronze.sql_scalar_func(p_par INT)
RETURNS INT 
LANGUAGE SQL
return p_par*10
```

```python
%sql
select total_amount , databricksansh.bronze.sql_scalar_func(total_amount) from databricksansh.silver.sales_enr
```

### **PYTHON**

```python
%sql
CREATE OR REPLACE FUNCTION databricksansh.bronze.python_scalar_function(p_par INT)
RETURNS INT
LANGUAGE PYTHON
AS 
$$
    return p_par*100 
$$
```

```python
%sql
select total_amount , databricksansh.bronze.python_scalar_function(total_amount) from databricksansh.silver.sales_enr
```
